- Ejercicio guiado 2: detector básico de spam
- Objetivo: recorrer una lista de palabras sospechosas, contar coincidencias y clasificar el mensaje según su puntaje de riesgo.

```python
mensaje = input("Escribe un mensaje: ").lower()

palabras_sospechosas = ["gratis", "premio", "ganaste", "urgente", "haz clic", "contraseña"]
riesgo = 0  

for palabra in palabras_sospechosas:
    if palabra in mensaje:
        riesgo += 1

print("Puntaje de riesgo:", riesgo)
```

```python
if riesgo == 0:
    print("Mensaje normal")
elif riesgo <= 2:
    print("Mensaje sospechoso")
else:
    print("Posible spam")
```
