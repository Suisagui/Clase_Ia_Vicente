## Preparación
Este archivo usa una copia didáctica del dataset Iris con **8 celdas vacías introducidas para practicar limpieza**. Son 150 flores, tres especies y cuatro mediciones. Los datos originales se guardan en `iris_original.csv` para comparar.

Coloca este notebook en la misma carpeta que los archivos CSV. Si aparece `FileNotFoundError`, revisa la carpeta abierta en VS Code.

```python
import pandas as pd

iris = pd.read_csv("iris_con_faltantes.csv")
print("Dataset cargado:", iris.shape)
display(iris.head(3))
```

**Comprobación esperada:** `(150, 5)`.

**Referencia de columnas:** `sepal_length_cm`, `sepal_width_cm`, `petal_length_cm`, `petal_width_cm`, `species`. Las medidas están expresadas en centímetros.

## Ejercicio 1. Explorar el dataset

a) Muestra las primeras cinco filas. b) Identifica las dimensiones. c) Consulta los tipos de columna y sus valores no nulos. d) Encuentra cuáles especies aparecen en la columna `species`.

**Pista:** Recuerda `head`, `shape`, `info` y `unique`. Debes encontrar **3 especies**.


```python
display(iris.head())
print("Dimensiones:", iris.shape)
iris.info()
print("Especies:", iris["species"].unique())
```

**Respuesta:** Hay **150 filas y 5 columnas**. Las cuatro mediciones son numéricas (float64) y species contiene texto. Las especies son setosa, versicolor y virginica. Hay 145 valores registrados en sepal_width_cm, 147 en petal_width_cm y 150 en las demás columnas.

## Ejercicio 2. Seleccionar columnas y posiciones

a) Crea una tabla que contenga solo `species` y `petal_length_cm`. b) Muestra **las cinco primeras filas y las dos primeras columnas** mediante posiciones, no nombres.

**Pista:** Usa una lista de columnas para (a) y `iloc` para (b). El final de un slice no se incluye.


```python
seleccion = iris[["species", "petal_length_cm"]]
display(seleccion)

primeras_posiciones = iris.iloc[:5, :2]
display(primeras_posiciones)
```

**Respuesta:** La primera tabla contiene solo la especie y la longitud del pétalo. iloc[:5, :2] selecciona las filas de posición 0 a 4 y las columnas de posición 0 y 1: longitud y ancho del sépalo.

## Ejercicio 3. Filtrar con una o dos condiciones

a) Selecciona solamente flores de la especie `setosa`. b) Selecciona flores de `virginica` cuyo `petal_length_cm` sea **mayor que 5.5**. c) Muestra cuántas filas tiene cada resultado.

**Pista:** Utiliza una máscara de comparación y `&`; cada condición debe ir entre paréntesis.


```python
setosa = iris[iris["species"] == "setosa"]
virginica_larga = iris[
    (iris["species"] == "virginica") & (iris["petal_length_cm"] > 5.5)
]

display(setosa)
display(virginica_larga)
print("Flores setosa:", len(setosa))
print("Virginica con pétalos mayores que 5.5 cm:", len(virginica_larga))
```

**Respuesta:** Hay **50 flores setosa** y **25 virginica** con pétalos de más de 5.5 cm. La comparación es estricta: los pétalos que miden exactamente 5.5 cm no entran en el segundo resultado.

## Ejercicio 4. Contar y agrupar por especie

a) Cuenta cuántas flores hay de cada especie. b) Calcula el **promedio de `petal_length_cm` para cada especie**. c) Identifica cuál tiene el promedio mayor.

**Pista:** Busca `value_counts`, `groupby` y `mean`. La columna `petal_length_cm` no contiene faltantes en esta copia.


```python
conteo = iris["species"].value_counts()
promedios = iris.groupby("species")["petal_length_cm"].mean()

print("Flores por especie:")
print(conteo)
print("\nPromedio de longitud del pétalo en cm:")
print(promedios)
print("\nEspecie con el promedio mayor:", promedios.idxmax())
```

**Respuesta:** Cada especie tiene **50 flores**. El promedio de longitud del pétalo es **1.462 cm** para setosa, **4.260 cm** para versicolor y **5.552 cm** para virginica. El promedio mayor corresponde a **virginica**.

## Ejercicio 5. Limpiar valores faltantes y quitar columnas

a) Cuenta cuántos valores faltan por columna. b) Crea **una copia** sin las filas que tengan valores vacíos y cuenta cuántas quedan. c) Crea otra copia donde rellenes únicamente los faltantes de `sepal_width_cm` con su mediana. d) Crea una tercera copia sin la columna `petal_width_cm`. **No modifiques `iris` directamente.**

**Pista:** Recuerda `isna().sum()`, `dropna()`, `fillna()`, `median()` y `drop(columns=[...])`. Esperamos **142 filas completas** antes de rellenar.


```python
print("Faltantes por columna:")
print(iris.isna().sum())

iris_sin_faltantes = iris.dropna().copy()
print("\nFilas completas:", len(iris_sin_faltantes))

iris_rellenado = iris.copy()
mediana_ancho_sepalo = iris["sepal_width_cm"].median()
iris_rellenado["sepal_width_cm"] = iris_rellenado["sepal_width_cm"].fillna(
    mediana_ancho_sepalo
)
print("\nMediana del ancho del sépalo:", mediana_ancho_sepalo)
print("Faltantes después del relleno:")
print(iris_rellenado.isna().sum())

iris_sin_ancho_petalo = iris.drop(columns=["petal_width_cm"]).copy()
print("\nColumnas de la tercera copia:", iris_sin_ancho_petalo.columns.tolist())
print("Faltantes en iris, sin modificar:", iris.isna().sum().sum())
```

**Respuesta:** Faltan **5 valores de ancho del sépalo** y **3 de ancho del pétalo**. Al eliminar las filas incompletas quedan **142**. La mediana del ancho del sépalo es **3.0 cm**; al rellenar solo esa columna quedan 3 faltantes, todos en el ancho del pétalo. La copia sin petal_width_cm tiene 4 columnas. Son tres alternativas separadas e iris conserva sus 150 filas, 5 columnas y 8 faltantes.

## Ejercicio 6. Transformar columnas

a) Crea en una copia una nueva columna `petal_length_mm` convirtiendo centímetros a milímetros. b) Crea otra columna `species_codigo` asignando `S`, `V` y `G` a las tres especies. c) Comprueba cinco filas.

**Pista:** Para la conversión basta una operación sobre toda la columna; para códigos puedes usar `.map()` con un diccionario. Si quieres, prueba también `.apply(lambda ...)`.


```python
iris_transformado = iris.copy()
iris_transformado["petal_length_mm"] = iris_transformado["petal_length_cm"] * 10

codigos = {"setosa": "S", "versicolor": "V", "virginica": "G"}
iris_transformado["species_codigo"] = iris_transformado["species"].map(codigos)

display(iris_transformado.head())
```

**Respuesta:** Se multiplica por **10** para pasar de centímetros a milímetros: por ejemplo, 1.4 cm son 14 mm. Se asigna S a setosa, V a versicolor y G a virginica. La copia transformada tiene **150 filas y 7 columnas**; la tabla iris sigue sin cambios.

## Ejercicio 7. Exportar el resultado

a) Guarda el DataFrame transformado del ejercicio 6 en `iris_procesado.csv`, **sin guardar la columna de índice**. b) Vuelve a leer el CSV con Pandas y comprueba que conserva 150 filas y las nuevas columnas.

**Pista:** Usa `to_csv` con `index=False` y después `read_csv`.


```python
iris_transformado.to_csv("iris_procesado.csv", index=False)
iris_leido = pd.read_csv("iris_procesado.csv")

print("Dimensiones del CSV:", iris_leido.shape)
print("Columnas:", iris_leido.columns.tolist())
print("Conserva 150 filas:", len(iris_leido) == 150)
print("Conserva las columnas nuevas:",
      {"petal_length_mm", "species_codigo"}.issubset(iris_leido.columns))
print("Sin columna de índice adicional:",
      iris_leido.columns.tolist() == iris_transformado.columns.tolist())
display(iris_leido.head())
```

**Respuesta:** El archivo exportado tiene **150 filas y 7 columnas**, incluidas petal_length_mm y species_codigo, y no agrega una columna de índice. Conserva los 8 faltantes porque el ejercicio 6 transforma una copia de iris, no la copia rellenada del ejercicio 5.

## Ejercicio 8. Unir filas con `concat`

a) Construye dos DataFrames: primeras 50 filas y siguientes 50 filas de `iris`. b) Únelos verticalmente y comprueba el número total de filas.

**Pista:** Combina `iloc` y `pd.concat`; prueba `ignore_index=True`.


```python
primeras_50 = iris.iloc[:50].copy()
siguientes_50 = iris.iloc[50:100].copy()
iris_unido = pd.concat([primeras_50, siguientes_50], ignore_index=True)

print("Filas del primer grupo:", len(primeras_50))
print("Filas del segundo grupo:", len(siguientes_50))
print("Dimensiones de la unión:", iris_unido.shape)
print("Índice:", iris_unido.index.min(), "a", iris_unido.index.max())
display(iris_unido.head())
```

**Respuesta:** Las dos partes tienen 50 filas cada una. concat produce una tabla de **100 filas y 5 columnas**, con un índice nuevo de 0 a 99. Contiene las primeras 100 observaciones de iris, sin añadir las últimas 50.

## Ejercicio 9. Relacionar tablas con `merge`
Crea una tabla pequeña con las columnas `species` y `codigo`: setosa→S, versicolor→V y virginica→G. Une esta tabla con `iris` mediante la especie. Verifica que aparecen 150 filas y una columna adicional. Los códigos son solamente etiquetas didácticas.

**Pista:** Puedes construir la tabla auxiliar con `pd.DataFrame({...})` y hacer un `merge` por `species`.


```python
tabla_codigos = pd.DataFrame({
    "species": ["setosa", "versicolor", "virginica"],
    "codigo": ["S", "V", "G"]
})
display(tabla_codigos)

iris_con_codigos = iris.merge(tabla_codigos, on="species", how="left")
print("Dimensiones tras merge:", iris_con_codigos.shape)
print("Códigos sin asignar:", iris_con_codigos["codigo"].isna().sum())
display(iris_con_codigos.head())
display(iris_con_codigos[["species", "codigo"]].drop_duplicates())
```

**Respuesta:** La unión tiene **150 filas y 6 columnas**. Se añade codigo y no quedan códigos vacíos. No se duplican flores porque cada especie aparece una sola vez en la tabla auxiliar.
